```python
# ------------------- Read the file and assign column names -------------------------------------------
path = "/home/poornima/Documents/BSE_METISSE_project/output_binary/"
m1_str = f"{int(m1 * 100):05d}"
m2_str  = f"{int(m2 * 100):05d}"
filename = f"{m1_str}M_{m2_str}M.dat"        ## file_name = "00200M_00100M.dat"
full_path = os.path.join(path, file_name)

data = np.loadtxt(full_path) 

# Remove the last row
data = data[:-1, :]

# Define the column names
columns = [
    "Time",
    "stellar_type_1","stellar_type_2","initial_mass_1","initial_mass_2", "current_mass_1","current_mass_2","log10_r_1","log10_r_2",
    "R1/ROL1","R2/ROL2","log10_L_1","log10_L_2",
    "spin_1","spin_2","mass_loss_rate_1","mass_loss_rate_2","convective_envelope_mass_1","convective_envelope_mass_2",
    "envelope_radius_1","envelope_radius_2","sep_Rsun","ecc"]
    
df = pd.DataFrame(data, columns=columns)

#df.head()

# --------------- RLOF -------------------------------------------------------

# Primary star - see RLOF time and primary star's ev. state
mask_primary = df["R1/ROL1"] >= 1
if mask_primary.any():  # only if there is at least one True
    RLOF_time_primary = df["Time"][mask_primary].min()
    RLOF_ev_state_primary = df[df["Time"] == RLOF_time_primary]["stellar_type_1"].iloc[0]
    print(RLOF_time_primary,RLOF_ev_state_primary)
else:
    RLOF_time_primary = np.nan

# Secondary star - see RLOF time and secondary star's ev. state
mask_secondary = df["R2/ROL2"] >= 1
if mask_secondary.any():
    RLOF_time_secondary = df["Time"][mask_secondary].min()
    RLOF_ev_state_secondary = df[df["Time"] == RLOF_time_secondary]["stellar_type_2"].iloc[0]
    print(RLOF_time_secondary,RLOF_ev_state_secondary)
else:
    RLOF_time_secondary = np.nan

###print(df.iloc[1])

### see which star goes through RLOF first and make that star the donar star 
if RLOF_time_primary < RLOF_time_secondary :
    donar_state = RLOF_ev_state_primary
elif RLOF_time_primary > RLOF_time_secondary :
    donar_state = RLOF_ev_state_secondary
elif RLOF_time_primary == RLOF_time_secondary :
    donar_state = RLOF_ev_state_primary
elif np.isnan(RLOF_time_primary) and np.isnan(RLOF_time_secondary):
    donar_state = np.nan
elif np.isnan(RLOF_time_primary):
    donar_state = RLOF_ev_state_secondary
elif np.isnan(RLOF_time_secondary):
    donar_state = RLOF_ev_state_primary

#print(donar_state)

###  Assign case
if donar_state == 1.0:   # MS
    case = 0  # case A
elif donar_state == 2.0 or donar_state == 3.0 :  # HG or FGB
     case = 1 # case B[HG]
elif donar_state == 4.0:  # CHeB
     case = 2  # case B[CHeB]
elif donar_state == 5.0 or donar_state == 6.0:  # First Asymptotic Giant Branch or Second Asymptotic Giant Branch
     case = 3 # case C
elif np.isnan(donar_state) :
     case = 4 # None
else :
     case = 5 # any other state

f1.write(f"{m1} {p} {case} {donar_state}\n")    

# ---------------------------------------------12 Gyr -------------------------------------------------------------------------

subset_after_twelve_gyr = df[df["Time"] > 12000]  ### get subset where time is more than 12 Gyr


### Find the seperation after 12 Gyr
if not subset_after_twelve_gyr.empty:
    sep_12_gyr = float(subset_after_twelve_gyr["sep_Rsun"].iloc[0])
else:
    sep_12_gyr = np.nan  

### Find the type of stars after 12 Gyr
if not subset_after_twelve_gyr.empty:
    m1_type_now = subset_after_twelve_gyr["stellar_type_1"].iloc[0]
    m2_type_now = subset_after_twelve_gyr["stellar_type_2"].iloc[0]
else:
    m1_type_now = np.nan 
    m2_type_now = np.nan 

### Find the situation of the binaries after 12 Gyrs

if sep_12_gyr != 0.0 :
    if {m1_type_now, m2_type_now} == {14.0}: ### BBH
        last_state = 1
    elif {m1_type_now, m2_type_now} == {13.0, 14.0}: ### BH-NS
        last_state = 2
    elif {m1_type_now, m2_type_now} == {13.0} :     ###BNS
        last_state = 3
    elif ((m1_type_now in {10.0, 11.0, 12.0, 13.0, 14.0} and m2_type_now in {10.0, 11.0, 12.0}) or
        (m2_type_now in {10.0, 11.0, 12.0, 13.0, 14.0} and m1_type_now in {10.0, 11.0, 12.0})):      ### WD - Compact Object
        last_state = 4
    elif 0 <= int(m1_type_now) <= 9 and 0 <= int(m2_type_now) <= 9: ### non-double compact objects
        last_state = 5
    elif m1_type_now == 15 or m2_type_now == 15:    ### Massless Supernova
        last_state = 6
elif sep_12_gyr ==0.0 :     ### Mergers
    if {m1_type_now, m2_type_now} == {14.0}: ### BBH
        last_state = 7
    elif {m1_type_now, m2_type_now} == {13.0, 14.0}: ### BH-NS
        last_state = 8
    elif {m1_type_now, m2_type_now} == {13.0} :     ###BNS
        last_state = 9
    elif ((m1_type_now in {10.0, 11.0, 12.0, 13.0, 14.0} and m2_type_now in {10.0, 11.0, 12.0}) or
        (m2_type_now in {10.0, 11.0, 12.0, 13.0, 14.0} and m1_type_now in {10.0, 11.0, 12.0})):      ### WD - Compact Object
        last_state = 10
    elif 0 <= int(m1_type_now) <= 9 and 0 <= int(m2_type_now) <= 9: ### non-double compact objects
        last_state = 11
    elif m1_type_now == 15 or m2_type_now == 15:    ### Massless Supernova
        last_state = 12
else :
    last_state = 13

f2.write(f"{m1} {p} {sep_12_gyr} {m1_type_now} {m2_type_now} {last_state}\n")   
```
