```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import matplotlib.colors as colors
from matplotlib.ticker import LogLocator, LogFormatterMathtext
import numpy as np
import matplotlib as mpl
```

```python
data = pd.read_csv("./data_file_1.dat", sep=r"\s+")

fig, ax = plt.subplots(figsize=(3, 2))


case_colors = {
    0: '#9467bd',
    1: '#ffbb78',
    2: '#1f77b4',
    3: '#c49c94',
    4: '#c7c7c7'
    #5: 'green'
}

primary_mass = data['primary_mass']
period = data['period']

colors = data['case'].map(case_colors)
    
ax.scatter(primary_mass , period, color=colors, marker='.')

legend_elements = [
    Line2D([0], [0], marker='.', color='#9467bd',  linestyle='None', label='Case A'),
    Line2D([0], [0], marker='.', color='#ffbb78',  linestyle='None', label='Case B [HG]'),
    Line2D([0], [0], marker='.', color='#1f77b4',  linestyle='None', label='Case B[HeB]'),
    Line2D([0], [0], marker='.', color='#c49c94',  linestyle='None', label='Case C'),
    Line2D([0], [0], marker='.', color='#c7c7c7', linestyle='None', label='None'),
    
]

ax.legend(
    handles=legend_elements,
    loc='upper left',
    bbox_to_anchor=(1.02, 1.0),   # pushes legend outside
    borderaxespad=0,
    frameon=False
)

ax.set_xlabel(r'Mass of primary $/M_\odot$')        # initial (ZAMS) mass
ax.set_ylabel('Orbital period / days')      # final remnant mass
# ax.set_ylim(ymin=1, ymax=10000) ##10000
# ax.set_xlim(xmin=10, xmax=40) ##40

ax.set_box_aspect(1)

# X padding (linear)
ax.set_xlim(10 - 2, 40 + 2)

# Y padding (log scale)
ax.set_yscale('log')
ax.set_ylim(1 / 2, 10000 * 2)

ax.set_xticks([10, 20, 30, 40])
ax.set_yticks([1, 10, 100, 1000, 10000])

plt.show()

ax.set_title('Fig 13') 


img_title = "RLOF"

# save_name = output_img_dir/f"{img_title}.png"
save_name = f"{img_title}.png"

fig.savefig(save_name, bbox_inches="tight")
```

```python
data = pd.read_csv("./data_file_2.dat", sep=r"\s+")

fig, ax = plt.subplots(figsize=(3, 2))


case_colors = {
    1: '#c5b0d5', # BBH
    8: '#9467bd', # BBH Merger
    2: '#ffdd99', # BH-NS
    9: '#ff9900', # BH-NS Merger
    3: '#aec7e8',  # BNS
    10: '#1f77b4', # BNS Merger
    4: '#c49c94', # WD/BH/NS-WD
    11: '#8b0000',  # WD/BH/NS-WD Merger
    5: '#d3d3d3',  # Non DCO
    12: '#7f7f7f',   # Non DCO Merger  
    6: 'lightgreen',   # 15,15   
    13 : 'darkgreen'   # 15,15 merger
   
}

primary_mass = data['primary_mass']
period = data['period']

#colors = data['case'].map(case_colors)
colors = data['case'].map(case_colors).fillna('#000000')  
    
ax.scatter(primary_mass , period, color=colors, marker='.')

legend_elements = [
    # BBH
    Line2D([0], [0], marker='.', color='#c5b0d5', linestyle='None', label='BBH'),
    Line2D([0], [0], marker='.', color='#9467bd', linestyle='None', label='BBH merger'),

    # BH–NS
    Line2D([0], [0], marker='.', color='#ffdd99', linestyle='None', label='BH-NS'),
    Line2D([0], [0], marker='.', color='#ff9900', linestyle='None', label='BH-NS merger'),

    # BNS
    Line2D([0], [0], marker='.', color='#aec7e8', linestyle='None', label='BNS'),
    Line2D([0], [0], marker='.', color='#1f77b4', linestyle='None', label='BNS merger'),

    # WD/BH/NS–WD
    Line2D([0], [0], marker='.', color='#c49c94', linestyle='None', label='WD/BH/NS-WD'),
    Line2D([0], [0], marker='.', color='#8b0000', linestyle='None', label='WD/BH/NS-WD merger'),

    # non-DCO
    Line2D([0], [0], marker='.', color='#d3d3d3', linestyle='None', label='non-DCO'),
    Line2D([0], [0], marker='.', color='#7f7f7f', linestyle='None', label='non-DCO mergers'),

    # Massless Supernova
    Line2D([0], [0], marker='.', color='lightgreen', linestyle='None', label='Massless Supernova'),
    Line2D([0], [0], marker='.', color='darkgreen', linestyle='None', label='Massless Supernova Merger'),
]

ax.legend(
    handles=legend_elements,
    loc='upper left',
    bbox_to_anchor=(1.02, 1.0),   # pushes legend outside
    borderaxespad=0,
    frameon=False
)

ax.set_xlabel(r'Mass of primary $/M_\odot$')        # initial (ZAMS) mass
ax.set_ylabel('Orbital period / days')      # final remnant mass


ax.set_box_aspect(1)

# X padding (linear)
ax.set_xlim(10 - 2, 40 + 2)

# Y padding (log scale)
ax.set_yscale('log')
ax.set_ylim(1 / 2, 10000 * 2)

ax.set_xticks([10, 20, 30, 40])
ax.set_yticks([1, 10, 100, 1000, 10000])

plt.show()

ax.set_title('Fig 13') 


img_title = "12_Gyr_cases"

# save_name = output_img_dir/f"{img_title}.png"
save_name = f"{img_title}.png"

fig.savefig(save_name, bbox_inches="tight")


## Where should I put massless supernova case ?
```
