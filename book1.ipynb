# Plot HR diagram for selected sets of systems

```python
import json
import matplotlib.pyplot as plt
import matplotlib as mpl
from pathlib import Path
```

```python
output_img_dir = Path("./output_img_dir")
```

```python
fig, ax = plt.subplots(figsize=(7, 6))

# Logarithmic color normalization for stellar mass
norm = mpl.colors.LogNorm(vmin=10, vmax=40.0)
cmap = mpl.cm.ScalarMappable(norm=norm, cmap='viridis_r')
cmap.set_array([])

with open("new_data_3.jsonl") as f:
    for line in f:
        obj = json.loads(line)

        # Select only p = 1
        if obj["p"] == 1:

            m1 = obj["m1"]

             # Plot track with color determined by mass
            ax.plot(obj["log_T_eff_1"], obj["log_L_1"], color=cmap.to_rgba(m1))

            

# --- Add colorbar for stellar mass ---
cbar = plt.colorbar(cmap, ax=ax)
cbar.set_label(r'Stellar mass [$M_\odot$]')

# --- Proper log-scale ticks ---
ticks = [10, 15, 20, 25, 30, 35, 40]
cbar.set_ticks(ticks)
cbar.set_ticklabels([str(t) for t in ticks])


ax.set_xlabel(r'log T$_{\mathrm{eff}}$/K')
ax.set_ylabel(r'log L/L$_{\odot}$')
ax.invert_xaxis()
ax.set_title("Hertzsprung–Russell diagram")

img_title = "HR_diag_p1_m1"

save_name = output_img_dir/f"fig_{img_title}.png"

fig.savefig(save_name, dpi=300, bbox_inches="tight")

plt.show()
```
