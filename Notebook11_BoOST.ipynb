```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import os
from pathlib import Path
import pandas as pd
from IPython.display import clear_output
import time
import json
```

```python
def read_file(m1,m2):
    # ------------------- Read the file -------------------------------------------
    path = "/home/poornima/Documents/BSE_METISSE_project/output_binary/"
    m1_str = f"{int(m1 * 100):05d}"
    m2_str  = f"{int(m2 * 100):05d}"
    file_name = f"{m1_str}M_{m2_str}M.dat"        ## file_name = "00200M_00100M.dat"
    full_path = os.path.join(path, file_name)

    return (full_path)

def make_df(full_path):
    data = np.loadtxt(full_path) 

    # Remove the last row
    data = data[:-1, :]
    
    # Define the column names
    columns = [
        "Time",
        "stellar_type_1","stellar_type_2","initial_mass_1","initial_mass_2", "current_mass_1","current_mass_2","log10_r_1","log10_r_2",
        "R1/ROL1","R2/ROL2","log10_L_1","log10_L_2",
        "spin_1","spin_2","mass_loss_rate_1","mass_loss_rate_2","convective_envelope_mass_1","convective_envelope_mass_2",
        "envelope_radius_1","envelope_radius_2","sep_Rsun","ecc"]
        
    df = pd.DataFrame(data, columns=columns)

    return(df)

```

```python
def count_lines(df,f1,m1,m2,p):

    num_of_lines = len(df)

    time_max = df["Time"].iloc[-2]
    time_3rd_last = df["Time"].iloc[-3]
    time_4th_last = df["Time"].iloc[-4]

    sep_at_maxtime = df["sep_Rsun"].iloc[-2]
    ecc_at_maxtime = df["ecc"].iloc[-2]

    selected_cols = ["sep_Rsun", "ecc", "R1/ROL1", "R2/ROL2"]

    has_nan = df[selected_cols].isna().any().any()
    

    f1.write(f"{m1} {m2} {p} {num_of_lines} {time_max} {time_3rd_last} {time_4th_last} {sep_at_maxtime} {ecc_at_maxtime} {has_nan}\n")  

    return ()
    
```
